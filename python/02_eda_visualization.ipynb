# Exploratory Data Analysis (EDA)

## Objective

This notebook performs exploratory data analysis on the loan approval dataset to understand the distribution of important applicant and loan-related variables and identify patterns associated with loan approval.

The analysis focuses on:

- Loan approval distribution
- Credit Score and loan approval
- Annual Income and loan approval
- Loan Amount Requested and loan approval
- Employment Status and loan approval

The findings from this analysis will help understand the key factors associated with loan approval and guide the machine learning stage.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df= pd.read_csv(r"C:\Users\USER\OneDrive\Desktop\Loan-Credit-Analytics\dataset\Loan_Dataset.csv")
```

```python
df.columns = df.columns.str.strip()
print(df.columns.tolist())
```

```python
sns.countplot(x="Loan_Approval_Status", data=df)
plt.title("Loan Approval Distribution")
plt.xlabel("Loan Approval Status")
plt.ylabel("Number of Applicants")
plt.show()
```

```python
sns.boxplot(x="Loan_Approval_Status", y="Credit_Score", data=df)
plt.title("Credit Score vs Loan Approval")
plt.xlabel("Loan Approval Status")
plt.ylabel("Credit Score")
plt.show()
```

```python
sns.boxplot(x="Loan_Approval_Status", y="Annual_Income", data=df)
plt.title("Annual Income vs Loan Approval")
plt.xlabel("Loan Approval Status")
plt.ylabel("Annual Income")
plt.show()
```

```python
sns.boxplot(
    x="Loan_Approval_Status",
    y="Loan_Amount_Requested",
    data=df
)

plt.title("Loan Amount vs Loan Approval")
plt.xlabel("Loan Approval Status")
plt.ylabel("Loan Amount Requested")
plt.show()
```

```python
sns.countplot(
    x="Employment_Status",
    hue="Loan_Approval_Status",
    data=df
)

plt.title("Employment Status vs Loan Approval")
plt.xlabel("Employment Status")
plt.ylabel("Number of Applicants")
plt.xticks(rotation=20)
plt.show()
```
